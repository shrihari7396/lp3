```python
x = 2
learning_rate = 0.1
precision = 0.00001
previous_step_size = 1
max_iters = 10000
iters = 0

df = lambda x: 2 * (x + 3)

while previous_step_size > precision and iters < max_iters:
    prev_x = x
    x = x - learning_rate * df(prev_x)
    previous_step_size = abs(x - prev_x)
    iters += 1

print(f"Local minimum occurs at x = {x:.5f}")
print(f"Minimum function value y = {(x + 3)**2:.5f}")
```
