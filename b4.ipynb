```python
# Gradient Descent for y = (x + 3)^2

x = 2
learning_rate = 0.1
iterations = 100

for i in range(iterations):
    y = (x + 3) ** 2
    gradient = 2 * (x + 3)

    print(f"Iteration {i + 1}: x = {x:.4f}, y = {y:.4f}, Gradient = {gradient:.4f}")

    x = x - learning_rate * gradient

print(f"\nMinimum x = {x:.4f}")
print(f"Minimum y = {(x + 3) ** 2:.4f}")
```
