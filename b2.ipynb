```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import scale
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn import metrics
```

```python
df = pd.read_csv('emails.csv')
df.head()
```

```python
df = df.dropna()

X = scale(df.drop(columns=['Email No.', 'Prediction']))
Y = df['Prediction']
```

```python
knn_model = KNeighborsClassifier(n_neighbors=7)
x_train, x_test, y_train, y_test = train_test_split(X, Y, random_state=42, test_size=0.3)
```

```python
knn_model.fit(X=x_train, y=y_train)
knn_model_pred = knn_model.predict(X=x_test)


print("\n--- KNN Results ---")
print("Accuracy:", metrics.accuracy_score(y_test, knn_model_pred))
print("Precision:", metrics.precision_score(y_test, knn_model_pred))
print("Recall:", metrics.recall_score(y_test, knn_model_pred))
print("F1 Score:", metrics.f1_score(y_test, knn_model_pred))
print("\nClassification Report:\n",
      metrics.classification_report(y_test, knn_model_pred))
```

```python
svm_model = SVC(C=1)
svm_model.fit(x_train, y_train)
svm_model_predict = svm_model.predict(x_test)

print("\n--- SVM Results ---")
print("Accuracy:", metrics.accuracy_score(y_test, svm_model_predict))
print("Precision:", metrics.precision_score(y_test, svm_model_predict))
print("Recall:", metrics.recall_score(y_test, svm_model_predict))
print("F1 Score:", metrics.f1_score(y_test, svm_model_predict))
print("\nClassification Report:\n",
      metrics.classification_report(y_test, svm_model_predict))
```
